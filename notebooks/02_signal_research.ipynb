# 02 - Signal research

Construct the locked 12-2 signal and verify exposure and temporal integrity. No strategy parameter is selected from the post-2000 test.

```python
from pathlib import Path

from industry_momentum.data import prepare_data
from industry_momentum.features import assert_temporal_integrity, cross_sectional_weights, momentum_signal

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
returns = prepare_data(ROOT / 'data' / 'raw').industry_returns
signals = momentum_signal(returns, lookback_months=12, skip_months=1)
weights = cross_sectional_weights(signals, selection_fraction=0.20)
assert_temporal_integrity(returns, lookback_months=12, skip_months=1)
```

```python
active = weights.abs().sum(axis=1) > 0
exposure = weights.loc[active].agg(['sum'])
assert (weights.loc[active].clip(lower=0).sum(axis=1).round(12) == 1).all()
assert (weights.loc[active].clip(upper=0).sum(axis=1).round(12) == -1).all()
weights.loc[active].head()
```
