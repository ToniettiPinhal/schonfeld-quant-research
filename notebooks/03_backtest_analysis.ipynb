# 03 - Backtest analysis

Run the primary portfolio and inspect the locked test. The complete grid, bootstrap, leave-one-out analysis, figures, and research note are generated by `python scripts/run_research.py`.

```python
from pathlib import Path

from industry_momentum.backtest import run_strategy
from industry_momentum.config import ResearchConfig
from industry_momentum.data import prepare_data
from industry_momentum.metrics import performance_metrics

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
config = ResearchConfig(raw_data_dir=ROOT / 'data' / 'raw', reports_dir=ROOT / 'reports')
bundle = prepare_data(config.raw_data_dir)
result = run_strategy(bundle.industry_returns, config.strategy)
```

```python
test = result.monthly.loc[config.test_start:config.test_end]
performance_metrics(test['net_return'], turnover=test['traded_notional'])
```

```python
(1 + test[['gross_return', 'net_return']]).cumprod().plot(logy=True, title='Post-2000 industry momentum')
```
